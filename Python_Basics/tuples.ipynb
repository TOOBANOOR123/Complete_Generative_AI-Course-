## Tuples:
Tuples are ordered collections of items that are immutable. they are similar to lists, but their immutability makes them different. 

## Creating a Tuple

```python
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
#for creating an empty tuple
tple = tuple()
print(type(tple))
#for empty list 
lst = list()
print(type(lst))
```

```python
numbers = tuple([1,2, 3, 4, 5, 5, 6]) #a list can be converted into tuple
print(numbers)
```

```python
lst = list(numbers) ## also tuples can be converted into list
print(lst)
```

```python
mixed_tuple = (1, "Hellow World", 3.14, 38)
print(mixed_tuple)
```

## Accessing elements from the tuple

## Immutable nature of tuples
-  meaning that their elements cannot be changed once assigned

```python
lst = [1,2,3,4,5]
print(lst)
##now as the lists are mutable/chnageable we can change the elements once assigned
lst[1] = "Krish"
print(lst)
```

```python
# tuples are immutable
mixed_tuple[2] = "Genai"
print(mixed_tuple)
```

there are the ways to change the elements in the tuple. first by assigning the tuple as list the, change the elements and then convert back the list into tuple

## Packing and Unpacking tuple

```python
packed_tuple = 1,3.14,"hellow", "Green"
print(packed_tuple, type(packed_tuple)) ##whenever we assign the elements like this without braces and brackets, it will packed them into a tuple saying packing a tuple
```

```python
## unpacking the tuple
a, b, c, d = packed_tuple
print(a)
print(b)
print(c)
print(d)
```

```python
# Unpacking with *
first, *middle, last = packed_tuple
#star means from where it is starting upack the elements till last(excluding)
print(first)
print(middle)
print(last)
```

## Nested tuple 


```python
#nested list 
lst = [[2,3,1], ["Aplle", "Orange", "Peach"], ["Red", "Orange", "Peach"], [1, "Hellow World", 3.15, "c"]]
lst[0][2] 
```

```python
lst[0]
```

```python
lst[3]
```

```python
#inside a list we can also have tuple
#nested list 
lst = [[2,3,1], ["Aplle", "Orange", "Peach"], ["Red", "Orange", "Peach"], (1, "Hellow World", 3.15, "c")]
lst[3]
```

```python
##nested tuple 
nested_tuple = ((2,3,4), ("ab", "bc", "fd", "th"), (True, False))
print(nested_tuple)
print(nested_tuple[0][0])
print(nested_tuple[1][0:])
print(nested_tuple[2])
```

## Iterating over the nested tuples

```python
for sub_tup in nested_tuple:
    for item in sub_tup:
        print(item, end = " ")
    print() 
```
