## Iterating over the list

```python
numbers = [1,2, 3, 4, 5, 6, 6,7, 8, 9 , 10]
for number in numbers:
    print(number)
```

```python
fruits_names = ["Apple", "Orange", "Banana", "Kiwi"]
for fruits in fruits_names:
    print(fruits)
```

## iterating elements of the list with respect to its indices

```python
numbers = [1,2, 3, 4, 5, 6, 6,7, 8, 9 , 10]
for number in enumerate(numbers):
    print(number)
```

```python
for index, number in enumerate(numbers):
    print(index, number)
```

```python
for fruits in enumerate(fruits_names):
    print(fruits)

for index, fruits in enumerate(fruits_names):
    print(index, fruits)
```

## List Comprehension

- Basics Syntax [expression for item in iterable]
- with conditional logic [expression for item in iterable if condiiton]
- Nested List Comprehension [expression for item1 in iterable1 for item2 in iterable2]

```python
lst = []
for x in range(10):
    lst.append(x**2)
print(lst)
```

#### Basic Syntax

```python
square_lst = [x**2 for x in range(10)]
print(square_lst)
```

##### List Comprehension with conditon

```python
lst = []
for i in range(10):
    if i%2 == 0:
        lst.append(i)
print(lst)
```

```python
even_numbers = [num for num in range(29) if num%2 == 0]
print(even_numbers)
```

```python
odd_numbers = [num for num in range(20) if num%2 != 0]
print(odd_numbers)
```

### Nested list comprehension

```python
lst1 = ["Apple", "Orange", "Banana", "Mango", "Strawberry"]
lst2 = ["Red", "Orange", "Yellow", "Yellow", "Pink"]

pair = [[i,j] for i in lst1 for j in lst2]
print(pair)
```

## list comprehesnion witj function calls

```python
lst1 = ["Apple", "Orange", "Banana", "Mango", "Strawberry"]
lengths = [len(word) for word in lst1]
print(lengths)
```
