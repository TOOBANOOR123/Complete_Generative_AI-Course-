# Loops in Python

## 1.Introduction to loops
     A loop lets your program repeat a block of code multiple times without rewriting it.
     Think of it like telling a computer:
         "Do this task again and again --- until I say stop"
     Loops save time, reduce mistakes, and make programs more powerful.

```python
range(5)
```

# iterate numbers with for loop

```python
for i in range(1, 5):
    print(i)
```

```python
range(1, 20, 2) #range is tarting from 1 to 20 with the step size 2
```

```python
for i in range(1, 20, 2):
    print(i)
```

```python
for i in range(1, 20, -1):
    print(i)
```

```python
for i in range(20, 1, -1):
    print(i)
```

```python
##strings 
str = "Mr. Chips"
for i in str:
    print(i)
```

## While Loop

   The while loop continues to execute as long as the conditon is True.

```python
count= 0 
while count < 10:
    print(count)
    count = count+1
```

```python
#While loop 
# the while loop continues to execute as long as the conditon is True
count = 0 
while count < 5:
    print(count)
    count = count+1
```

```python
count = 0 
while count%2 == 0:
    print(count)
    count = count+1
```

# Break statement

```python
for i in range(10):
    if i == 3:
        break #exiting the loop permanently with break statement
    print(i)


```

## Continue statement 

```python
for i in range(10):
    if i%2 == 0:
        continue ##skips the loop iteration on the following condition
    print(i)
```

## Pass statement 

```python
#pass is a null operation and it does nothing
for i in range(10):
    if i == 3:
        pass
    print(i)
```

## Nested Loop

```python
for i in range(3):
    for j in range(2):
        print(f"i: {i} and j: {j}")
```

## Examples:  Calculate the sum of first N natural numbers using a while and for loop


#### While loop 

```python
n = 10 
sum = 0
count = 1

while count <= n:
    sum = sum+count 
    count = count+1

print("Sum of first 10 natural number:", sum)
```

#### For Loop

```python
n = 10
sum = 0
count = 1
for i in range(1, n+1):
    sum = sum+i
    i = i+1

print("Sum of first 10 natural numbers: ", sum)
```

## Example2: Display Prime numbers between 1 and 100

```python
for num in range(1, 101):
    if num>1:
        for i in range(2, num):
            if num%i == 0:
                break
        else:
            print(num)
```
