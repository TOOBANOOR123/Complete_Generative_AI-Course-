## If else statement

```python
age = int(input("Enter your Age: "))
if age>20:
    print("Age is greater than 20")
else: 
    print("Age is not greater than 20")
```

## Number even, odd, negative

```python
num = int(input("Enter the number: "))
if num>=0:
    print("the number is positive")
    if num%2==0:
        print("the number is even.")
    else: 
        print("the number is odd.")
else:
    print("The number is negative")
```

## Simple Calculator

```python
num1 = float(input("Enter First Number: "))
num2 = float(input('Enter second number: '))
operation = input("Enter Operation(+, -, *, /): ")

if operation == "+":
    print("Sum of Two numbers is: ", int(num1+num2))
elif operation == "-":
    print("Difference between two numbers: ", num1-num2)
elif operation == "*":
    print("Product of both numbers is: ", num1*num2)
elif operation == "/":
    if num2 == 0:
        print("Error: Division by zero is not allowed!")
    else:
        print("Quotient of two numbers is:", num1/num2)
else:
    print("Invalid Operation.")
```

## Determine if a year is a leap year using nested conditon statement

```python
year = int(input("Enter the year: "))

if year%4 == 0:
    if year%100 == 0:
        if year%400 == 0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
else: 
    print(year, "is not a leap year")
```

## Determine the ticket price based on age abd whether the person is a student
