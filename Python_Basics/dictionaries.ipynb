# Introduction ot Dictionaries
Dictionaries are an unordered collection of items. They store data in key-value pairs. keys must be unique and immuatble (e.g.m strings, numbers, or tuple), while values can of any type.

## Creating a dictionary

```python
dic = dict()
print(type(dic))
```

```python
empty_dict = {}
print(empty_dict, type(empty_dict))
```

```python
student = {"name":"Tooba", "age":26, "Gender":"Female", "Height":5.2}
print(student)
```

```python
print(type(student))
```

```python
#dictionary keys must be unique what will happened if there are duplicated keys, it will just print first ones not the others.
student = {"name":"Tooba", "age":26, "Gender":"Female", "Height":5.2, "name":"Ayesha", "Age":26, "Height":5.6, "Gender": "Female"}
print(student)
```

```python
# Accessing Dictionary Elements
print(student['name'])
print(student['age'])

##Accessing using get() method
print(student.get('age'))
print(student.get("name"))
```

```python
print(student.get('surname'))
```

```python
##can give the default value when the value is not available
print(student.get('surname', 'Noor'))
```

## Modifying Dictionary elements

```python
student['age'] == 25
student['name'] == 'Amina'
```

```python
student['age'] = 25 ##updating the key 
student['name'] = 'Amina' #updating the key 
##adding the new key value
student['surname'] = 'Noor'
```

```python
print(student.get('age'))
print(student.get('name'))
print(student.get('surname'))
```

```python
##deleting the key 
del student['surname'] ##this will delete the key value pair
print(student)
```

```python
## Dictionary methods
keys = student.keys() ##get all keys
print(keys)
values = student.values() #get all values
print(values)

##get all key value pairs 
items = student.items() #get all key value pairs
print(items)
```

## Shallow Copy

```python
## Shallow copy 
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "Tuba"
print(student)
print(student_copy)
```

```python
#make a separate copy
student_copy1 = student.copy() ##it will allocate the differenet memory-shallow copy
print(student_copy1)
print(student)
```

```python
student["name"] = "Tuba"
print(student)
print(student_copy1)
```

## iterating over dictionaires
 you can use loops to iterate over dictionaires, keys, values, or items
 

```python
##iterating over keys
for keys in student.keys():
    print(keys)
```

```python
#iterating over dictionaries values 
for value in student.values():
    print(value)
```

```python
##iterating over key value pairs
for keys, values in student.items():
    print(f'{keys}:{values}')
```
