# Inheritence in Python
inheritence is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class.

```python
## Inheritence 
## Parent Class
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    def drive(self):
        print(f"The person will drive the {self.enginetype} car")
```

```python
car1 = Car(4, 4, "Petrol")
print(car1)
```

```python
car1.drive()
```

```python
print(car1.windows)
print(car1.doors)
```

```python
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, is_self_driving):
        super().__init__ (windows, doors, enginetype)
        self.is_self_driving = is_self_driving
    def selfdriving(self):
        if self.is_self_driving:
            print("The car is self driving")
        else:
            print("the Car is not self driving")
```

```python
car2 = Tesla(4, 4, "Electric", True)
```

```python
print(car2.windows)
```

```python
print(car1.enginetype)
print(car2.enginetype)
```

```python
car1.drive()
car2.drive()
```

```python
car2.selfdriving()
```

#### Multiple Inheritence
when a class inherits from more than one base class.

```python
## Base Class 1
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species
    def speak(self):
        if self.species == "Dog":
            return "woof woof"
        elif self.species == "Cat":
            return "meow meow"
        elif self.species == "Cow":
            return "moo moo"
        elif self.species == "Duck":
            return "quack quack"
        elif self.species == "Lion":
            return "roar raor"
        elif self.species == "sheep":
            return "baa baa"
        elif self.species == "Horse":
            return "neigh neigh"
        elif self.species == "Monkey":
            return "ooh ooh aah aah"
        elif self.species == "Donkey":
            return "hee haw"
        else:
            return "Unknown sound"
        
##Base class 2
class Pet:
    def __init__(self, owner): 
        self.owner = owner 
## Derived class
class sub_animal(Animal, Pet):
    def __init__(self, name, species,owner, breed):
        Animal.__init__(self, name, species)
        Pet.__init__(self, owner)
        self.breed = breed
    def display_info(self):
        return f"{self.name} is a {self.breed} breed and is owned by {self.owner} and says {self.speak()}"
```

```python
dog1 = sub_animal("Buddy", "Dog", "Alice", "Golden Retriver")
print(dog1.display_info())
```

```python
cat1 = sub_animal("Whiskers", "Cat", "Bob", "Siamese")
print(cat1.display_info())
cat1.speak()
```

```python
lion = sub_animal("Simba", "Lion", "Charlie", "African Lion")
print(lion.display_info())
print(lion.owner)
lion.speak()
```
