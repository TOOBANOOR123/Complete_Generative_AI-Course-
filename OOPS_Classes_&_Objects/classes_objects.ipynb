## Classes & Objects:
Object-Oriented Programming (OOP) is a programming paradigm that uses "objects" to design applications and computer programs. OOP allows for modelling real-world scenarios using classes and objects. 

```python
### A class is a blue print for creating objects. It defines a set of attributes and methods that the objects created from the class will have.
class Car:
    pass
# We can create an object from the class by calling the class name followed by parentheses.
my_car = Car()
audi = Car()

print(my_car)  # Output: <__main__.Car object at 0x7f8b8c8c8c8>
print(audi)  # Output: <__main__.Car object at 0x7f8b8c8c8c8>
print(type(my_car))  # Output: <class '__main__.Car'>
print(type(audi))  # Output: <class '__main__.Car'>
```

```python
## instance variables and methods
my_car.windows = 4
my_car.color = "red"
print(my_car.windows)  # Output: 4
print(my_car.color)  # Output: red
```

```python
### instance variables and methods are specific to the object and can be different for different objects created from the same class.
class Cat:
    ##constructor
    def __init__(self, name, age):
        self.name = name 
        self.age = age 
cat1 = Cat("Butterpuff", 1)
cat2 = Cat("Whiskers", 3)

print(cat1)
print(cat2)
```

```python
print(cat1.name)
print(cat2.age)
```

```python
##Define a class with instance variables and methods
class Dog:
    def __init__(self, name, age):
        self.name = name 
        self.age = age 
    def bark(self):
        print(f"{self.name} say woof!")
    

dog1 = Dog("buddy", 2)
dog2 = Dog("max", 4)
print(dog1)
dog1.bark()
print(dog2)
dog2.bark()
```

```python
## Modeling a Bank Account
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number 
        self.balance = balance 
    def deposit(self, amount):
        self.balance += amount
        print(f"{self.account_number} deposited {amount}. New Balance: {self.balance}")
    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Insufficient funds in account {self.account_number}")
        else:
            self.balance -= amount 
            print(f"{self.account_number} withdrew {amount}. New Balance is {self.balance}")
    def get_balance(self):
        return self.balance
```

```python
##lets create an object of bank account
account1 = BankAccount("12345667", 50000)
print(account1)
```

```python
account1.deposit(10000)
print(f"Current Balance after depositing amount: {account1.get_balance()}")
account1.withdraw(10000)
account1.withdraw(20000)
account1.withdraw(15000)
print(f"Current Balance: {account1.get_balance()}")
```
