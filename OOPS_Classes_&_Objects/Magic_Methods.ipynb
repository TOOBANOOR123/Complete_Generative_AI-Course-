# Magic Methods
Magic methods in Python, also known as dunder methods (doublle underscore methods), are special methods that start and end with double underscore. These methods enable you to define the behaviours of objects for built-in operations, such as arithmetic operation, comparisons, and more.

## Magic Methods
Magic Methods are predefined methods in Python that you can override to change the behaviour of your objects. Some common magic methods include:

```python
## Basic Methods

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

per = Person("Alice", 30)
print(per.name)  # Output: Alice
print(per)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name 
        self.age = age 
    ##overiding the __str__ method to return a string representation of the object
    def __str__(self):
        return f"Person(name = {self.name}, age = {self.age})"
    
per = Person("Alice", 30)
print(per)
```
