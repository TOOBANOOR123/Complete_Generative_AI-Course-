```python
corpus = "Hi, My name is Tooba. I am learning krish Naik's GenAI course. it is fun to learn."
```

```python
print(corpus)
```

```python
from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
```

```python
print(documents)
print(type(documents))
```

```python
for sentence in documents:
    print(sentence)
```

```python
from nltk.tokenize import word_tokenize
```

```python
words = word_tokenize(corpus)
print(words)
```

```python
for word in words:
    print(word)
```

```python
print(type(word))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
opunc = wordpunct_tokenize(corpus)
```

```python
print(opunc)
```

```python
for punc in opunc:
    print(punc)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tri = TreebankWordTokenizer()
```

```python
tri.tokenize(corpus)
```
