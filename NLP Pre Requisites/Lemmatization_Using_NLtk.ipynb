## Lemmatization
Lemmatization technique is like stemming. The output we get after lemmatization is called "lemma", which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting the valid word that means the same thing. NLTK provides WordNetLemmatizer Class which is a thin wrapper around the wordnet corpus. Tis class uses morphy() function to the wordnet CorpusReader class to find a lemma.

#### Examples
-------- Q/A Chatbot 
-------- Text Summarization 

```python
from nltk.stem import WordNetLemmatizer
```

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
### Parts of Speech
'''Noun-n
verb-v
adjective-a
adverb-r'''
```

```python
lemmatizer.lemmatize("going", pos='r')
```

```python
words = ["eating", "eat", "going", "working", "finding", "ate"]
```

```python
for word in words: 
   lem = lemmatizer.lemmatize(word, pos = 'a')
   print(lem)
```

```python
for word in words:
    lem = lemmatizer.lemmatize(word, pos='r')
    print(lem)
```

```python
for word in words:
    lem = lemmatizer.lemmatize(word, pos='v')
    print(lem)
```

```python
lemmatizer.lemmatize("eating", pos='r')
```

```python
lemmatizer.lemmatize("eating", pos='n')
```

```python
lemmatizer.lemmatize("eating", pos='v')
```
